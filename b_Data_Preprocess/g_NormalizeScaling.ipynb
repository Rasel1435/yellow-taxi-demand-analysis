## **Q/A**

### **What is Normalization / Scaling?**

Normalization (or Scaling) is the process of transforming numerical feature values into a common range or scale so that no single feature dominates the others.

In simple words:

- It makes all features comparable by putting them on a similar scale.

Example before scaling:
```bash
Income: 20,000 – 200,000
Age: 18 – 60
```
Income will dominate the model unfairly. Scaling fixes this.

### **Why do we use Normalization / Scaling?**

We use it to:

1. Prevent Feature Dominance
    Large values overpower small values.
2. Improve Model Accuracy
    Models learn more balanced patterns.
3. Speed up Training
    Especially for gradient-based algorithms.
4. Improve Distance Calculations

Crucial for:

- KNN
- K-Means
- SVM
- PCA
- Neural Networks

Not strictly necessary for:

- Decision Trees
- Random Forest
- XGBoost

5. Stabilize Optimization

Helps gradient descent converge faster.

### **Normalization (Min-Max Scaling)**

Scales data between 0 and 1.

**Example:**

If values go from 0 to 100:

```bash
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

Formula: X_scaled = X - X_min / X_max - X_min

Original: 50
Scaled: 0.5
```
Best for:

- When data has fixed min & max
- Image processing
- Neural networks



### **Standardization (Z-score Scaling)**

Scales data to mean = 0 and std dev = 1.

```bash
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

Formula: Z = X − μ​ / σ

Original: 50
Scaled: 0.5
```
Best for:

- When data follows normal distribution
- Linear Regression, SVM, PCA
